```python
import pandas as pd
import matplotlib.pyplot as plt
import pdb
```

```python
# Load the two CSV files
df1 = pd.read_csv("D:/MyDrive/Stability/Transect_2017.csv")
df2 = pd.read_csv("D:/MyDrive/Stability/Transect_2022.csv")

df1["distance_scaled"] = df1["distance"] * 111
df2["distance_scaled"] = df2["distance"] * 111

# Plot both on the same figure
fig, ax = plt.subplots(figsize=(15,5))
ax.plot(df1["distance_scaled"], df1["SAMPLE_1"], label="2017", color="red",linewidth=2)
ax.plot(df2["distance_scaled"], df2["SAMPLE_1"], label="2022", color="black",linewidth=2)
low, mid, high = 8.6e-06, 2.4e-05, 8e-05

# Add horizontal patches (y spans)
plt.axhspan(0, low, facecolor="orange", alpha=0.1, label="Bottomfast")
plt.axhspan(low, mid, facecolor="blue", alpha=0.1, label="Stabilized")
plt.axhspan(mid, high, facecolor="black", alpha=0.1, label="Nonstabilzied")

x_target = 0.3
row = df1.iloc[(df2["distance_scaled"] - x_target).abs().argmin()]
x_val = row["distance_scaled"]
y_val = row["SAMPLE_1"]

# Add marker at the point
#plt.plot(x_val, y_val, marker="o", color="black", markersize=6)

# Safe annotation arrow
plt.annotate(
    f"Tide Crack",           # text label
    xy=(x_val, y_val+0.000001),        # point to annotate
    xycoords='data',
    xytext=(-15, 30),          # offset in points
    textcoords='offset points',
    arrowprops=dict(facecolor='black', arrowstyle='->', lw=1.5),
    fontsize=14
)



x_target = 7.2
row = df2.iloc[(df2["distance_scaled"] - x_target).abs().argmin()]
x_val = row["distance_scaled"]
y_val = row["SAMPLE_1"]

# Add marker at the point
#plt.plot(x_val, y_val, marker="o", color="black", markersize=6)

# Safe annotation arrow
plt.annotate(
    f"Barrier\nIsland",           # text label
    xy=(x_val, y_val+0.00002),        # point to annotate
    xycoords='data',
    xytext=(-20, 50),          # offset in points
    textcoords='offset points',
    arrowprops=dict(facecolor='black', arrowstyle='->', lw=1.5),
    fontsize=14
)

x_target = 13.3
row = df1.iloc[(df2["distance_scaled"] - x_target).abs().argmin()]
x_val = row["distance_scaled"]
y_val = row["SAMPLE_1"]

# Add marker at the point
#plt.plot(x_val, y_val, marker="o", color="black", markersize=6)

# Safe annotation arrow
plt.annotate(
    f"Grounded Ridge",           # text label
    xy=(x_val, y_val+0.00001),        # point to annotate
    xycoords='data',
    xytext=(-20, 50),          # offset in points
    textcoords='offset points',
    arrowprops=dict(facecolor='black', arrowstyle='->', lw=1.5),
    fontsize=14
)

x_target1 = 26.2
row = df2.iloc[(df2["distance_scaled"] - x_target1).abs().argmin()]
x_val = row["distance_scaled"]
y_val = row["SAMPLE_1"]

# Add marker at the point
#plt.plot(x_val, y_val, marker="o", color="black", markersize=6)

# Safe annotation arrow
plt.annotate(
    f"Grounded Ridge",           # text label
    xy=(x_val, y_val+0.000001),        # point to annotate
    xycoords='data',
    xytext=(-20, 50),          # offset in points
    textcoords='offset points',
    arrowprops=dict(facecolor='black', arrowstyle='->', lw=1.5),
    fontsize=14
)
ax.tick_params(axis='both', labelsize=14)
ax.yaxis.get_offset_text().set_fontsize(14)
ax.set_ylim([0, 0.00008])
ax.set_xlim([-1, 50])
ax.set_xlabel("Distance Along Transect (km)",fontsize=14)
ax.set_ylabel(r"Apparent Strain ($\epsilon_a$)",fontsize=14)
fig.text(0.97, 0.14, "C", fontsize=24, fontweight='bold', ha='right', va='bottom')
plt.legend(loc="upper right",fontsize=14)
plt.tight_layout()

plt.show()
fig.savefig("D:/MyDrive/Stability/Figs/transect_appstrain.png", dpi=300, bbox_inches="tight") 
```
