```python
from pathlib import Path
import datetime

# Folder with your files
folder = Path(r"D:\MyDrive\Stability\RawData\Monthly_Averages\MergedOutput_month_year\BOEM_GeoTIFFs")

for file in folder.rglob("i*_Coh.tif"):  # rglob makes it recursive
    # Example: i1201_1230_2024_Coh.tif
    name = file.stem  # i1201_1230_2024_Coh
    parts = name.split("_")
    if len(parts) != 4:
        print(f"Skipping unexpected format: {file.name}")
        continue

    first, second, year, _ = parts  # i1201, 1230, 2024, Coh
    year = int(year)

    # Extract MMDD
    mmdd1 = first[1:]  # skip the 'i'
    mmdd2 = second

    # Convert to DOY
    doy1 = datetime.datetime.strptime(f"{year}{mmdd1}", "%Y%m%d").timetuple().tm_yday
    doy2 = datetime.datetime.strptime(f"{year}{mmdd2}", "%Y%m%d").timetuple().tm_yday

    # Build new name
    new_name = f"i{year}_{doy1:03d}_{doy2:03d}_slie.tif"

    # Rename
    new_path = file.with_name(new_name)
    file.rename(new_path)
    print(f"Renamed: {file.relative_to(folder)} → {new_name}")
```
